```python
import pandas as pd
from scipy import stats

df = pd.read_csv("q4_team_shooting_raw.csv")
df.head()
```

```python
df['stage'].value_counts()
```

```python
sample_df = df.sample(n=30, random_state=42)
sample_df[['Squad', 'Sh', 'shots_per_match', 'stage']]
```

```python
sample_df.groupby('stage')['shots_per_match'].describe()
```

```python
n = len(sample_df)
mean = sample_df['shots_per_match'].mean()
std_err = stats.sem(sample_df['shots_per_match'])
ci = stats.t.interval(confidence=0.95, df=n-1, loc=mean, scale=std_err)
print(f"Sample mean: {mean:.3f}")
print(f"95% CI: ({ci[0]:.3f}, {ci[1]:.3f})")
```

```python
group_sh = sample_df[sample_df['stage'] == 'group_stage']['shots_per_match']
knockout_sh = sample_df[sample_df['stage'] == 'knockout']['shots_per_match']

t_stat, p_value = stats.ttest_ind(group_sh, knockout_sh, equal_var=False)

print(f"Group-stage mean: {group_sh.mean():.3f}")
print(f"Knockout mean: {knockout_sh.mean():.3f}")
print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("Reject H0: significant difference in shots per match between the two groups.")
else:
    print("Fail to reject H0: no significant difference detected.")
```
