```python
import pandas as pd
from scipy import stats

df = pd.read_csv("q3_ball_possession_raw.csv", skiprows=1)
df['Squad'] = df['Squad'].str.split(' ', n=1).str[1]
df = df.rename(columns={'Poss':'Possession'})

knockout_teams = [
    "Mexico", "South Africa", "Switzerland", "Canada", "Brazil", "Morocco",
    "USA", "Australia", "Germany", "Côte d'Ivoire", "Netherlands", "Japan",
    "Belgium", "Egypt", "Spain", "Cabo Verde", "France", "Norway",
    "Argentina", "Austria", "Colombia", "Portugal", "England", "Croatia",
    "Bosnia–Herz", "Paraguay", "Ecuador", "Sweden", "Senegal", "Algeria",
    "Congo DR", "Ghana"
]
df['stage'] = df['Squad'].apply(lambda x: 'knockout' if x in knockout_teams else 'group_stage')

df.head()
```

```python
df['stage'].value_counts()
```

```python
sample_df = df.sample(n=30, random_state=42)
sample_df[['Squad', 'Possession', 'stage']]
```

```python
sample_df.groupby('stage')['Possession'].describe()
```

```python
n = len(sample_df)
mean = sample_df['Possession'].mean()
std_err = stats.sem(sample_df['Possession'])
ci = stats.t.interval(confidence=0.95, df=n-1, loc=mean, scale=std_err)
print(f"Sample mean: {mean:.3f}")
print(f"95% CI: ({ci[0]:.3f}, {ci[1]:.3f})")
```

```python
group_poss = sample_df[sample_df['stage'] == 'group_stage']['Possession']
knockout_poss = sample_df[sample_df['stage'] == 'knockout']['Possession']

t_stat, p_value = stats.ttest_ind(group_poss, knockout_poss, equal_var=False)

print(f"Group-stage mean: {group_poss.mean():.3f}")
print(f"Knockout mean: {knockout_poss.mean():.3f}")
print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("Reject H0: significant difference in possession between the two groups.")
else:
    print("Fail to reject H0: no significant difference detected.")
```
